# Advanced Algorithm Tumbling

This notebook demonstrates advanced features of the Algorithm Tumbler system.

```python
from algorithms.core import *
from algorithms.evolution import EvolutionarySearch
from algorithms.library import ComponentLibrary

# Load a pre-built library
library = ComponentLibrary.load('data/seed_algorithms.json')

# Initialize evolutionary search
search = EvolutionarySearch(library, seed=42)
search.initialize(50)  # Population of 50

print(f"Library has {len(library.components)} components")
```

```python
# Define a complex task: find maximum in array
test_cases = [
    ([1, 3, 2, 5, 4], 5),
    ([10, 9, 8, 7, 6], 10),
    ([], None),  # Edge case
]

def fitness_fn(program, tests):
    score = 0
    for inp, expected in tests:
        try:
            result = program.execute({'input': inp})
            if result == expected:
                score += 10
        except:
            pass
    return score - program.ast_size() * 0.1  # Penalize complexity

# Run evolution
search.evolve(generations=100, fitness_fn=fitness_fn, test_cases=test_cases)

# Show best program
best = max(search.hall_of_fame, key=lambda p: fitness_fn(p, test_cases))
print(f"Best fitness: {fitness_fn(best, test_cases)}")
print(f"Program: {best}")
```
